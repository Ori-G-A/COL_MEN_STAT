```python
df = df[['AÑO', 'DEPARTAMENTO', 'POBLACIÓN_5_16', 'TASA_MATRICULACIÓN_5_16', 'COBERTURA_NETA', 'COBERTURA_BRUTA', 'REPROBACIÓN', 'REPITENCIA', 'DESERCIÓN']]
```

## Análisis unidimensional de las variables

```python
fig, axes = plt.subplots(4, 2, figsize=(16, 24))
sns.histplot(
    data=df,
    x='POBLACIÓN_5_16',
    bins=50,
    kde=True,
    ax=axes[0, 0],
    color = 'olive'
)
axes[0, 0].set_title('Distribución del Ingreso Mensual')

#2. Histograma fecha de vinculación
sns.histplot(
    data=df,
    x='TASA_MATRICULACIÓN_5_16',
    bins=50,
    kde=True,
    ax=axes[0, 1],
    color = 'olive'
)
axes[0, 1].set_title('Distribución de la fecha de vinculación')

sns.histplot(
    data=df,
    x='COBERTURA_NETA',
    bins=50,
    kde=True,
    ax=axes[1, 0],
    color = 'salmon'
)
axes[1, 0].set_title('Distribución del Ingreso Mensual')

#2. Histograma fecha de vinculación
sns.histplot(
    data=df,
    x='COBERTURA_BRUTA',
    bins=50,
    kde=True,
    ax=axes[1, 1],
    color = 'salmon'
)
axes[1, 1].set_title('Distribución de la fecha de vinculación')


sns.histplot(
    data=df,
    x='REPROBACIÓN',
    bins=50,
    kde=True,
    ax=axes[2, 0],
    color = 'olive'
)
axes[2, 0].set_title('Distribución del Ingreso Mensual')

#2. Histograma fecha de vinculación
sns.histplot(
    data=df,
    x='REPITENCIA',
    bins=50,
    kde=True,
    ax=axes[2, 1],
    color = 'olive'
)
axes[2, 1].set_title('Distribución de la fecha de vinculación')

sns.histplot(
    data=df,
    x='DESERCIÓN',
    bins=50,
    kde=True,
    ax=axes[3, 0],
    color = 'salmon'
)
axes[3, 0].set_title('Distribución del Ingreso Mensual')
```
